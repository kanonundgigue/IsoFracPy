```python
def main(ISO_TYPE: str = "HDO"):
    temp_sea_list = [  0,   0,   0,  0, 0] # Initial sea surface tempratures
    temp_air_list = [-20, -15, -10, -5, 0]  # Initial air temperatures
    h_air = 0.6
    surface_wind = 30
    plot_effective_humidity(
    temp_air_list, 
    temp_sea_list,
    h_air = h_air
    )

    init_dict = get_init_conditions(
    temp_sea_list,
    temp_air_list,
    h_air = h_air,
    surface_wind = surface_wind,
    ISO_TYPE = ISO_TYPE
    )

    plot_init_conditions(init_dict, ISO_TYPE=ISO_TYPE)
```
